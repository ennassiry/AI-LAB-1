```python
import re
import nltk
import spacy
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
nltk.download("punkt")
nltk.download("stopwords")
```

```python
print("Partie 2 — Nettoyage du texte")
text = """
Bonjour ! Je suis étudiant à Marrakech.
J'aime beaucoup le NLP, Python et l'Intelligence Artificielle...
#NLP @Python :)
"""
text_lower = text.lower()
text_clean = re.sub(r"[^\w\s]", "", text_lower)
text_clean = re.sub(r"\s+", " ", text_clean).strip()
print(text_clean)

```

```python
# Partie 3 — Tokenisation et Stopwords

print("Partie 3 — Tokenisation et Stopwords")
tokens = word_tokenize(text_clean, language="french")
print("\nTokens :")
print(tokens)
stop_words = set(stopwords.words("french"))
filtered_tokens = []
for word in tokens:
    if word not in stop_words:
        filtered_tokens.append(word)

print("\nTokens apres suppression des stopwords :")
print(filtered_tokens)
```

```python
# Partie 4 — Lemmatisation
nlp = spacy.load("fr_core_news_sm")
filtered_text = " ".join(filtered_tokens)
print("\nTexte avant lemmatisation :")
print(filtered_text)
doc = nlp(filtered_text)
lemmas = [token.lemma_ for token in doc]
print("\nListe finale des lemmes :")
print(lemmas)
```

```python
text_finale = " ".join(lemmas)
print("# text finale")
print(text_finale)
```
